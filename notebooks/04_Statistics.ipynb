### Do late deliveries receive lower customer ratings than on-time deliveries?

### Null Hypothesis (H₀)

- The average customer rating is the same for late and on-time deliveries.

- Alternative Hypothesis (H₁)

- The average customer rating is different for late and on-time deliveries.

```python
import numpy as np

df['delivery_status'] = np.where(
    df['delivery_delay_days'] > 0,
    'late',
    'on_time'
)
```

```python
df[['delivery_delay_days','delivery_status']]
```

```python
alpha = 0.05
```

```python
late = df[df["delivery_status"] == "late"]["customer_rating"]

on_time = df[df["delivery_status"] == "on_time"]["customer_rating"]
```

```python
late
```

```python
from scipy.stats import ttest_ind
t_stat, p_value = ttest_ind(late, on_time)
```

```python
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

```python
from scipy.stats import f_oneway

groups = [
    group['net_amount'].values
    for _, group in df.groupby('category')
]

f_stat, p_value = f_oneway(*groups)

print(f"F-statistic: {f_stat:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("Average revenue differs across product categories.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("No significant difference in average revenue across categories.")
```
