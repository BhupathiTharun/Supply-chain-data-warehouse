### convert data types

```python
#change the data_type

od[['date_id', 'order_date', 'expected_delivery_date',
    'actual_delivery_date', 'return_date']] = (
    od[['date_id', 'order_date', 'expected_delivery_date',
        'actual_delivery_date', 'return_date']]
    .apply(pd.to_datetime)
)
```

```python
od.info()
```

```python
od.head()
```

```python
#check null or missing values
od.isna().sum()
```

### Handling Null values

```python
od[['seller_rating','delivery_rating','return_date','customer_rating','review_text']]
```

```python
#filling null values with no comments
od['review_text']=od['review_text'].fillna("No Comments")
```

```python
# rating columns is filled with mean
rating_cols = ['seller_rating', 'delivery_rating', 'customer_rating']

for col in rating_cols:
    od[col] = od[col].fillna(od[col].mean())
```

### Round Values

```python
# rating rounded
rating_cols = ['seller_rating', 'delivery_rating', 'customer_rating']

od[rating_cols] = od[rating_cols].round(1)
```

```python
od['seller_rating'].unique()
```

```python
od['net_amount']=od['net_amount'].round(2)
```

```python
od['return_date']=od['return_date'].fillna(np.nan)
```
